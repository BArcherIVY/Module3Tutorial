```python
class Solution:
    def sort012(self, arr):
        zero = 0
        one = 0
        two = 0

        for num in arr:
            if num == 0:
                zero += 1
            elif num == 1:
                one += 1
            elif num == 2:
                two += 1

        index = 0

        for i in range(zero):
            arr[index] = 0
            index += 1

        for i in range(one):
            arr[index] = 1
            index += 1

        for i in range(two):
            arr[index] = 2
            index += 1
```
